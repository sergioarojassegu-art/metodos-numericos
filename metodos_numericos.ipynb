## Cálculo de 1/x con precisión de 4 cifras significativas

Este programa calcula los valores de la función `1/x` de forma iterativa, comenzando en `x=1` y con un incremento de `0.0001` para `x`. Se detiene cuando el error relativo entre la `1/x` actual y la `1/x` anterior es menor que `0.0005`, lo que equivale a una precisión de 4 cifras significativas.

```python
x = 1
incremento = 0.0001
iteracion = 0
anterior = 1 / x

while True:
    x += incremento
    actual = 1 / x

    error_relativo = abs((actual - anterior) / actual)

    if error_relativo < 0.0005:
        break

    anterior = actual
iteracion += 1
print("\nResultado final:")
print(f"x = {x}")
print(f"Error relativo = {error_relativo:.8f}")
print(f"Iteraciones = {iteracion}")
```

```python
##falsa posicion
import numpy as np

def f(x):
    return x - np.cos(x)

a = -4
b = 3
tol = 0.5e-5
lista_c = []
lista_errores = []
i = 0
error = 1

while error > tol:
    f_a = f(a)
    f_b = f(b)
    c = c = a - f_a * (b - a) / (f_b - f_a)  # fórmula de falsa posición
    lista_c.append(float(c))

    if i > 0:
        error = abs((lista_c[i] - lista_c[i-1]) / lista_c[i])
        lista_errores.append(error)

    f_c = f(c)
    if f_a * f_c < 0:
        b = c
    else:
        a = c
    i += 1

print(lista_c)
print(lista_errores)
print("Cantidad de iteraciones:", i)
```

```python
##newton raphson

import numpy as np

def f(x):
    return x - np.cos(x)

def df(x):
    return 1 + np.sin(x)

x0 = -4  # valor inicial (puedes ajustarlo)
tol = 0.5e-5
lista_x = []
lista_errores = []
i = 0
error = 1

x = x0
while error > tol:
    x_nuevo = x - f(x) / df(x)  # fórmula de Newton-Raphson
    lista_x.append(x_nuevo)

    if i > 0:
        error = abs((lista_x[i] - lista_x[i-1]) / lista_x[i])
        lista_errores.append(error)

    x = x_nuevo
    i += 1

print(lista_x)
print(lista_errores)
print("Cantidad de iteraciones:", i)
```

```python
##secante

import numpy as np

def f(x):
    return x - np.cos(x)

x0 = -4
x1 = 3
tol = 0.5e-5
lista_x = [x0, x1]
lista_errores = []
i = 0
error = 1

while error > tol:
    f_x0 = f(lista_x[i])
    f_x1 = f(lista_x[i+1])
    x_nuevo = lista_x[i+1] - f_x1 * (lista_x[i+1] - lista_x[i]) / (f_x1 - f_x0)
    lista_x.append(float(x_nuevo))

    error = abs((lista_x[i+2] - lista_x[i+1]) / lista_x[i+2])
    lista_errores.append(float(error))

    i += 1

print(lista_x)
print(lista_errores)
print("Cantidad de iteraciones:", i)
```

```python
a = 1
b = 6
n = 6  # cifras de precisión deseadas

def area_trapecio(trap):
    puntos_x = trap + 1
    x = np.linspace(a, b, puntos_x)
    y = f(x)
    area_total = 0
    for i in range(trap):
        base = x[i+1] - x[i]
        altura = (y[i] + y[i+1]) / 2
        area_total = area_total + base * altura
    return area_total

trap = 1
A_anterior = area_trapecio(trap)
error = 1  # valor inicial grande para que entre al while

while error >= 0.5 * 10**(-n):
    trap += 1
    A_actual = area_trapecio(trap)
    error = abs(A_actual - A_anterior) / A_actual
    A_anterior = A_actual

print("Trapecios necesarios:", trap)
print("Área aproximada:", A_anterior)
print("Error final:", error)
```

```python
import numpy as np

def f(x):
    return 2*x[0]*x[1] + 2*x[0] - x[0]**2 - 2*x[1]**2

n = int(input("Número de cifras decimales de precisión: "))
tolerancia = 0.5 * 10**(-n)

R = ((5**0.5) - 1) / 2
x_opt = np.array([0.0, 0.0])

for i in range(100):
    x_anterior = x_opt.copy()

    for j in range(2):
        x0 = 0.0
        x3 = 4.0

        d = (x3 - x0) * R
        x1 = x3 - d
        x2 = x0 + d

        for k in range(100):
            punto1 = x_opt.copy()
            punto2 = x_opt.copy()
            punto1[j] = x1
            punto2[j] = x2

            if f(punto1) > f(punto2):
                x3 = x2
                x2 = x1
                d = (x3 - x0) * R
                x1 = x3 - d
            else:
                x0 = x1
                x1 = x2
                d = (x3 - x0) * R
                x2 = x0 + d

            if (x3 - x0) / 2 < tolerancia / 10:
                break

        x_opt[j] = (x0 + x3) / 2

    error = np.max(np.abs(x_opt - x_anterior))
    val_opt = f(x_opt)

    print(f"Iteración {i + 1}: x = {x_opt[0]}, "
          f"y = {x_opt[1]}, f(x,y) = {val_opt}, "
          f"cambio = {error}")

    if error < tolerancia:
        break
```
